```python
datasets = {
    "validations": validations,
    "routes": routes,
    "vehicles": vehicles,
    "stops": stops
}

for name, df in datasets.items():
    print("\nDataset:", name)
    
    null_profile = pd.DataFrame({
        "null_count": df.isna().sum(),
        "null_percentage": (df.isna().mean() * 100).round(2)
    })
    
    display(null_profile)
    
```

```python
categorical_columns = {
    "validations": ["fare_type", "transfer_flag"],
    "routes": ["mode", "district"],
    "vehicles": ["vehicle_type", "depot"],
    "stops": ["zone", "has_shelter"]
}

for name, columns in categorical_columns.items():
    print("\n==========", name, "==========")
    
    df = datasets[name]
    
    for col in columns:
        print("\nColumn:", col)
        print(df[col].value_counts(dropna=False))
```

## Data Quality Findings

The raw data contains several quality issues:

- Categorical values have inconsistent casing (e.g. Adult, adult, ADULT).
- Some text columns contain leading and trailing whitespace.
- Missing values exist in several columns.
- Duplicate validation records may exist.
- Some boolean fields use different formats (Y/N, 1/0, TRUE/FALSE).


```python
import os

os.makedirs("../data/interim", exist_ok=True)

validations.to_csv("../data/interim/validations_raw.csv", index=False)
routes.to_csv("../data/interim/routes_raw.csv", index=False)
vehicles.to_csv("../data/interim/vehicles_raw.csv", index=False)
stops.to_csv("../data/interim/stops_raw.csv", index=False)

print("Raw files saved successfully")
```
